```python
import os
from tkinter import *
from tkinter import ttk
import requests


def data_get():
    city = city_name.get()
    
    API_KEY = os.getenv("OPENWEATHER_API_KEY")
    data = requests.get(
        "https://api.openweathermap.org/data/2.5/weather",
        params={"q": city, "appid": API_KEY}
    ).json()
   
    w_lable_1.config(text=data["weather"][0]["main"])
    wb_lable_1.config(text=data["weather"][0]["description"])
    temp_lable_1.config(text=str(int(data["main"]["temp"]-272.15)))
    per_lable_1.config(text=data["main"]["pressure"])
    
win = Tk()

win.title("My Wheather App")
win.config(bg = "red")
win.geometry("500x570")
name_lable = Label(win, text = "My Wheather App", font = ("Time New Roman", 30, "bold"))
name_lable.place(x=25, y=50, height=50, width=450)

city_name = StringVar()

list_name = [
    "Karachi",
    "Lahore",
    "Faisalabad",
    "Rawalpindi",
    "Multan",
    "Gujranwala",
    "Hyderabad",
    "Khanewal",
    "Peshawar",
    "Islamabad",
    "Quetta",
    "Sargodha",
    "Sialkot",
    "Bahawalpur",
    "Sukkur",
    "Larkana",
    "Nawabshah",
    "Mirpur Khas",
    "Abbottabad",
    "Mardan",
    "Swat"
]

com = ttk.Combobox(win, text = "My Wheather App", values = list_name , 
                   font = ("Time New Roman", 20, "bold"), textvariable=city_name)
com.place(x=25, y=120, height=50, width=450)


w_lable = Label(win, text = "Wheather Climate", font = ("Time New Roman", 18))
w_lable.place(x=25, y=260, height=50, width=210)

w_lable_1 = Label(win, text = "", font = ("Time New Roman", 18))
w_lable_1.place(x=250, y=260, height=50, width=210)

wb_lable = Label(win, text = "Wheather Description", font = ("Time New Roman", 16))
wb_lable.place(x=25, y=330, height=50, width=210)

wb_lable_1 = Label(win, text = "", font = ("Time New Roman", 16))
wb_lable_1.place(x=250, y=330, height=50, width=210)

temp_lable = Label(win, text = "Temperature", font = ("Time New Roman", 18))
temp_lable.place(x=25, y=400, height=50, width=210)

temp_lable_1 = Label(win, text = "", font = ("Time New Roman", 18))
temp_lable_1.place(x=250, y=400, height=50, width=210)

per_lable = Label(win, text = "Pressure", font = ("Time New Roman", 18))
per_lable.place(x=25, y=470, height=50, width=210)

per_lable_1 = Label(win, text = "", font = ("Time New Roman", 18))
per_lable_1.place(x=250, y=470, height=50, width=210)


done_button = Button(win, text = "Done",  font = ("Time New Roman", 20, "bold"), command = data_get)
done_button.place(y = 190, height = 50, width = 100, x = 200)


win.mainloop()    
```
